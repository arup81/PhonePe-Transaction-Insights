# **Project Summary -**

This project focuses on analyzing digital transaction data to understand user behavior, transaction trends, and regional performance. The dataset includes information on transaction count, transaction amount, registered users, and app engagement across different states and districts over multiple years.

The data was first cleaned and preprocessed by handling missing values, removing duplicates, and merging multiple datasets into a single structured format. Exploratory Data Analysis (EDA) was then performed using various visualizations such as line charts, bar plots, scatter plots, histograms, and heatmaps to uncover patterns and relationships.

Key findings show that digital transactions have grown significantly over time, driven by an increasing number of users. However, the growth is uneven, with a few regions contributing a large portion of the total transaction value. Additionally, while user registration is high, user engagement remains relatively low in many areas.

The insights from this analysis suggest that businesses should focus on improving user engagement, targeting underperforming regions, and leveraging high-performing markets to achieve balanced and sustainable growth.

````python
df_state = pd.merge(df1, df2, 
                    on=['state', 'year', 'quarter'], 
                    how='outer')
````

````python
df_state = pd.merge(df_state, df3, 
                    on=['state', 'year', 'quarter'], 
                    how='outer')
````

````python
df_district = pd.merge(df4, df5, 
                       on=['state', 'district', 'year', 'quarter'], 
                       how='outer')
````

````python
df_pincode = pd.merge(df6, df7, 
                      on=['state', 'year', 'quarter'], 
                      how='outer')
````

````python
df = pd.concat([df_state, df_district, df_pincode], 
               ignore_index=True)
````

#### Chart - 1

````python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (10,6)
````

````python
time_data = df.groupby(['year','quarter'])['transaction_amount'].sum().reset_index()

plt.plot(range(len(time_data)), time_data['transaction_amount'], marker='o')
plt.title("Transaction Amount Growth Over Time")
plt.xlabel("Time (Year-Quarter Index)")
plt.ylabel("Transaction Amount")
plt.xticks(rotation=45)
plt.show()
````

##### 1. Why did you pick the specific chart?

A line chart is best for showing trends over time

##### 2. What is/are the insight(s) found from the chart?

#### Chart - 3

````python
district_df = df[df['district'] != 'State_Level']
top_districts = (
    district_df.groupby('district')['transaction_amount']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

fig, ax = plt.subplots()
sns.barplot(x=top_districts.values, y=top_districts.index, ax=ax)
ax.set_title("Top 10 Districts by Transaction Amount")
plt.show()
````
